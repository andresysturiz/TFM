📊 1. Cargar resultados

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

airfoil = pd.read_csv("../results/comparison_airfoil_self_noise.csv")
wine = pd.read_csv("../results/comparison_wine_quality.csv")
```

📈 2. Gráfico comparativo RMSE

```python
plt.figure(figsize=(10,5))
sns.barplot(data=airfoil, x="model", y="rmse")
plt.title("Comparación RMSE - Airfoil")
plt.xticks(rotation=45)
plt.show()
```

📈 3. Radar chart (opcional)

```python
import numpy as np

def radar_plot(df, title):
    labels = df["model"]
    stats = df["rmse"]

    angles = np.linspace(0, 2*np.pi, len(labels), endpoint=False)
    stats = np.concatenate((stats, [stats[0]]))
    angles = np.concatenate((angles, [angles[0]]))

    fig = plt.figure(figsize=(6,6))
    ax = fig.add_subplot(111, polar=True)
    ax.plot(angles, stats, "o-", linewidth=2)
    ax.fill(angles, stats, alpha=0.25)
    ax.set_thetagrids(angles[:-1] * 180/np.pi, labels)
    plt.title(title)
    plt.show()

radar_plot(airfoil, "Airfoil - RMSE Radar")
```

📉 4. Comparación PLS vs clásicos

```python
combined = pd.concat([
    airfoil.assign(dataset="Airfoil"),
    wine.assign(dataset="Wine")
])

sns.catplot(
    data=combined,
    x="model",
    y="rmse",
    hue="dataset",
    kind="bar",
    height=6,
    aspect=2
)
plt.xticks(rotation=45)
plt.title("Comparación RMSE entre datasets")
plt.show()
```
